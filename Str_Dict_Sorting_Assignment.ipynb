## Question 1. City Cleaner
A user types their city carelessly: " nEw dELhi ". Clean it and print exactly New Delhi. Also
validate that the city contains only alphabets and spaces — print Invalid city otherwise.

```python
city="  nEw   dELhi  "
city=" ".join(city.split()).title()
if all(ch.isalpha() or ch==" " for ch in city): # ch==" " for ch in city means allow spaces between words
    print(city)
else:
    print("Invalid city")
```

## Question 2. Prices Parser
Prices arrive as one messy string: " 120, 45 , 300 ,80, 95 ". Convert it into a list of integers, then
print the total bill, the highest price, and the average price with 2 decimal places (use an f-string with :.2f).

```python
prices=" 120, 45 , 300 ,80, 95 "
nums=[]
for p in prices.split(","):
    nums.append(int(p.strip()))
total=sum(nums)
print(f"Total   : {total}")
print(f"Highest : {max(nums)}")
print(f"Average : {total/len(nums):.2f}") # Average price with 2 decimal places
```

## Question 3. Username Validator
Validate a username entered by the user (e.g. " Rahul_123 "). Rules:
• clean the spaces and convert to lowercase
• must be alphanumeric (letters and digits only — no symbols)
• must start with an alphabet
• length must be between 5 and 12 characters
Print Valid username or Invalid username. ("Rahul_123" is invalid because of the underscore;
"rahul123" is valid.)

```python
username=" Rahul123 "
username=username.strip().lower()
if username.isalnum() and username[0].isalpha() and 5<=len(username)<=12:
    print("Valid username")
else:
    print("Invalid username")
```

## Question 4. Sentence Analysis
For a given sentence, print: (a) the total number of words, (b) how many words end with a vowel, and (c)
the longest word in the sentence.

```python
sentence="Python is easy to learn"
words=sentence.split()
v="aeiou"
count=0
longest=""
for w in words:
    if w[-1].lower() in v: # Check if the last character of a word is a vowel
        count+=1
    if len(w)>len(longest): # Check if the current word is longer than the longest word found so far
        longest=w
print("Words:",len(words))
print("Ending with vowel:",count)
print("Longest:",longest)
```

## Question 5. Character Frequency
Count the frequency of each character in a string, ignoring spaces and treating uppercase and lowercase
as the same letter. Example: "Aba c" → {'a': 2, 'b': 1, 'c': 1}.

```python
text="Aba c"
freq={} # Frequency dictionary
for ch in text.lower(): # Iterate through each character in the lowercase version of the text
    if ch!=" ":
        freq[ch]=freq.get(ch,0)+1 # Count the frequency of each character
print(freq)
```

## Question 6. Word Frequency
Count the frequency of each word in a sentence and print the word that occurs the most number of times
(use sorted() with key=lambda on the items, or a loop).

```python
s="cat dog cat bird dog cat"
freq={}
for w in s.split(): # Split the sentence into words and iterate through each word
    freq[w]=freq.get(w,0)+1
print(freq)
most=sorted(freq.items(),key=lambda x:x[1],reverse=True) # Sort the frequency dictionary by frequency
# key=lambda x: x[1] means we are sorting by the second element of each item (the frequency), reverse=True means we want the highest frequencies first
print("Most frequent:",most[0][0]) # most[0][0] gives the word and most[0][1] gives the frequency
```

## Question 7. Invert Dictionary
Invert the dictionary below so that values become keys. If two days share the same value, store both days
in a list:
sales = {"mon": 200, "tue": 350, "wed": 200}

```python
sales={"mon":200,"tue":350,"wed":200}
invert={}
for key,value in sales.items(): # key = day, value = sales amount, we have to invert and make value the key and key the value
    invert.setdefault(value,[]).append(key) # setdefault means if the key (value) doesn't exist, create an empty list
    # Append the day to the list of days having the same sales value
print(invert)
```

## Question 8. Merge Dictionaries
Marks of two tests are stored in two dictionaries. Merge them into one dictionary — if a student appears in
both, sum the mark.

```python
test1={"amit":40,"bina":35}
test2={"amit":30,"chirag":45}
result=test1.copy() # Create a copy of test1 to avoid modifying the original
for k,v in test2.items():
    if k in result:
        result[k]+=v
    else:
        result[k]=v
print(result)
```

## Question 9. Sort Employees
Sort the employees below by salary (highest first) using sorted() with key=lambda, and print each line
using an f-string in the format shown.

```python
employees=[("Ravi",55000),("Sita",72000),("Amit",48000)]
for k,v in sorted(employees,key=lambda x:x[1],reverse=True): # Sort by salary in descending order
    print(f"{k} earns Rs. {v}") # f-string for formatted output
```

## Question 10. Movies
Using the list of dictionaries below, 
• (a) Print the titles of all movies with rating > 8.1.
• (b) Extract all genres into a list (no duplicates).
• (c) Build a dict → {rating: [titles]} grouping movies by rating, then print it rating by rating using
f-strings, formatted nicely.

```python
movies=[
{"title":"Movie A","rating":8.2,"genre":"Thriller"},
{"title":"Movie B","rating":8.4,"genre":"Sports"},
{"title":"Movie C","rating":8.1,"genre":"Comedy"},
{"title":"Movie D","rating":8.2,"genre":"Drama"},
]
print("Ratings > 8.1")
for m in movies:
    if m["rating"]>8.1:
        print(m["title"])
genres=[]
for m in movies:
    if m["genre"] not in genres:
        genres.append(m["genre"])
print(genres)
group={}
for m in movies:
    group.setdefault(m["rating"],[]).append(m["title"]) # setdefault means "if the key (rating) doesn't exist, create it with a default value (empty list)"
for r in sorted(group):
    print(f"{r}: {group[r]}")
```
