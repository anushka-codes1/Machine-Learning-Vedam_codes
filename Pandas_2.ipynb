```python
import pandas as pd
import numpy as np

p = pd.read_csv("planets.csv")
t = pd.read_csv("Titanic-Dataset.csv")
tip = pd.read_csv("tips.csv")
```

```python
old_people = t.loc[t["Age"]>65,["Name","Pclass","Fare"]]
print(old_people)
```

```python
not_Southampton = t.loc[~(t["Embarked"]=="S")]
print("Not from Southampton: ",len(not_Southampton))
```

```python
under_10_or_over_70 = t.loc[(t["Age"]<10) | (t["Age"]>70)]
print(len(under_10_or_over_70))
```

```python
print(t["Pclass"].value_counts())
```

```python
males = t.loc[(t["Pclass"] == 3) & (t["Sex"] == "male")]
print("No. of Males in third class: ",len(males))
females = t.loc[(t["Pclass"] == 3) & (t["Sex"] == "female")]
print("No. of Females in third class: ",len(females))
```

```python
t["travel_group"] = "alone"
t.loc[(t["SibSp"]+t["Parch"]>0),"travel_group"] = "With family"
print(t["travel_group"])
```

```python
# p.dropna() removes every row with any blank in it. We have 1035 planets in planets.csv (p).
print(p.dropna()) # This does not modify p. It creates and returns a new DataFrame with rows containing missing values removed.
p.dropna(inplace=True) # This modifies p itself. Rows containing missing values are removed directly from p.
```

```python
# drop only where mass is blank
p.dropna(subset=["mass"])
```

```python
# drop the damages *columns* instead, and see which survive (axis = 1)
print(p.dropna(axis=1))
# drop the damages *rows* instead, and see which survive (axis = 0)
print(p.dropna(axis=0))
```

```python
print(p["mass"].fillna(0)) # filling all blank spaces in the mass column with 0
print(p["mass"].fillna(p["mass"].mean())) # filling all blank spaces in the mass column with the mean value of mass
print(p["mass"].fillna(p["mass"].median())) # filling all blank spaces in the mass column with the median value of mass
```

```python
p["mass"].fillna(0).isnull().sum()
```

```python
# How many exact duplicate rows?
print(p.duplicated().sum())
# Removing the duplicate rows
print(p.drop_duplicates())
```

```python
print(p.drop_duplicates(subset=["method"],ignore_index=True)) # ignore_index reorders the indexes again starting from 0 (ignoring the original indexes)
```
