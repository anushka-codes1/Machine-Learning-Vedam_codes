```python
import pandas as pd
bookings = pd.read_csv("bookings.csv")
```

```python
print(bookings['customer'].str.lower().str.strip())
print(bookings['customer'].str.lower().str.strip().unique()) # Due to the extra spaces b/w first name and last name, 
# duplicate values such as 'priya iyer' are also considered as unique values
```

```python
bookings['customer']=bookings['customer'].str.lower().str.strip().str.split().str.join(" ") # removing the extra spaces b/w first name and last name
print(bookings['customer'].unique())
```
