# Pandas Assignment — Selecting Data


## Setup

```python
import pandas as pd

# Load both datasets used in the assignment.
tips = pd.read_csv("tips.csv")
t = pd.read_csv("Titanic-Dataset.csv")
```

## A1

The owner wants to know how generous each customer was, not just how much they left. Make a new column called `tip_percent` - the tip as a percentage of the bill, rounded to 1 decimal place. Then show only `total_bill`, `tip` and `tip_percent` for the first 5 rows.

```python
# ---- A1 : your solution ----

# Calculate the tip as a percentage of the total bill.
tips["tip_percent"] = (tips["tip"] / tips["total_bill"] * 100).round(1)

# Select only the requested columns and show the first 5 rows.
tips[["total_bill", "tip", "tip_percent"]].head(5)
```

Answer both parts using positions only - no column names anywhere.

## A4

Two counts for the owner:

1. How many bills were not at Dinner? Use `~`.
2. How many bills had a party of 3 or more *and* a tip above 4?

```python
# ---- A4 : your solution ----

# ~ reverses the condition, so this selects rows where time is NOT Dinner.
not_dinner = tips.loc[~(tips["time"] == "Dinner")]
print("Bills not at Dinner:", len(not_dinner))

# & means both conditions must be true.
party_3_or_more_and_tip_above_4 = tips.loc[(tips["size"] >= 3) & (tips["tip"] > 4)]
print("Bills with party >= 3 and tip > 4:", len(party_3_or_more_and_tip_above_4))
```

## A5

The owner wants to flag the generous customers. Make a column called `service`. Everyone starts as `'normal'`. Any bill where `tip_percent` is more than 20 becomes `'generous'`. Then count how many of each. (Uses the column you made in A1.)

```python
# ---- A5 : your solution ----

# Everyone starts with the value 'normal'.
tips["service"] = "normal"

# Change only the rows where tip_percent is greater than 20.
tips.loc[tips["tip_percent"] > 20, "service"] = "generous"

# Count how many customers fall into each service category.
tips["service"].value_counts()
```

## B1

How many passengers boarded from each port? Use the `Embarked` column. Then add the three numbers up in a comment. Do they come to 891?

```python
# ---- B1 : your solution ----

# value_counts() counts the passengers for each recorded port.
port_counts = t["Embarked"].value_counts() # t is for Titanic-Dataset
print(port_counts)

# The three recorded port counts can be added together.
print("\nTotal of the three port counts:", port_counts.sum())

# Answer: No. The three counts add up to 889, not 891,
# because 2 passengers have a missing Embarked value.
```

## B2

Two counts:

1. How many women were travelling in third class?
2. How many of those women survived?

```python
# ---- B2 : your solution ----

# First condition: passenger is female AND passenger is in third class.
women_third_class = t.loc[(t["Sex"] == "female") & (t["Pclass"] == 3)]
print("Women travelling in third class:", len(women_third_class))

# From those women, select the passengers who survived.
women_third_class_survived = women_third_class.loc[women_third_class["Survived"] == 1]
print("Those women who survived:", len(women_third_class_survived))
```

## B3

Some people paid an enormous amount for their ticket. Show only `Name`, `Age` and `Fare` for every passenger who paid more than 200. How many are there?

```python
# ---- B3 : your solution ----

# Boolean indexing keeps only passengers whose Fare is greater than 200.
high_fare = t.loc[t["Fare"] > 200, ["Name", "Age", "Fare"]]

# Show the requested columns.
print(high_fare)

# Count the number of passengers matching the condition.
print("\nNumber of passengers who paid more than 200:", len(high_fare))
```

## B4

Out of all the passengers in first class, what percentage survived? Now do the same for third class. Put the two numbers next to each other in a comment.

```python
# ---- B4 : your solution ----

# Select all first-class passengers.
first_class = t.loc[t["Pclass"] == 1]

# Select all third-class passengers.
third_class = t.loc[t["Pclass"] == 3]

# Calculate the survival percentage for first class.
first_class_survival_percent = (first_class["Survived"].sum() / len(first_class)) * 100

# Calculate the survival percentage for third class.
third_class_survival_percent = (third_class["Survived"].sum() / len(third_class)) * 100

print("First-class survival percentage:", round(first_class_survival_percent, 2), "%")
print("Third-class survival percentage:", round(third_class_survival_percent, 2), "%")

# Answer: the two percentages are printed next to each other below.
```
