## Broadcasting

```python
# Broadcasting Rules:
# If the arrays have different numbers of dimensions, the smaller-dimensional array is "stretched" to match the larger-dimensional array.
# Only 1 can be stretched to any no. of rows or columns
# If the dimensions are different, padding (1) can be added to the left side of the smaller dimension array to make it have the same dimensions as the higher dimension array
# E.g 1: arr1 = np.array([10, 20, 30]) # 1,3 matrix
#      arr2 = np.array([[1], [2], [3]]) # 3,1 matrix
#      arr1 + arr2 will broadcast to a 3,3 matrix

# E.g. 2: arr3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]]) # 2,4,2 matrix
#      arr4 = np.array([[10], [20], [30], [40]]) # 4,1 matrix -> # 1,4,1 matrix (padding added to left side) -> stretched to # 2,4,2 matrix
#      arr3 + arr4 will broadcast to a 2,4,2 matrix
```

```python
# Broadcasting - 2D matrix + 1D row vector
arr2d = np.array([[1, 2, 3], [4, 5, 6]]) # 2,3 matrix
arr1d = np.array([10, 20, 30]) # 1,3 matrix (1 can be stretched to any number of rows or columns)
print("2D Array:")
print(arr2d)
print("1D Array:")
print(arr1d)
print("Result of Broadcasting:")
print(arr2d + arr1d) # This will broadcast to a 2,3 matrix
```

```python
# Broadcasting - row vector + column vector
row_vector = np.array([[10, 20, 30]])  # 1,3 matrix
column_vector = np.array([[1], [2], [3]])  # 3,1 matrix
print("Row Vector:")
print(row_vector)
print("Column Vector:")
print(column_vector)
print("Result of Broadcasting:")
print(row_vector + column_vector) # This will broadcast to a 3,3 matrix
```

```python
# reshape (rows,cols) - size must match
arr = np.arange(12).reshape(4,-1)
print("Reshaped Array:")
print(arr)
print(arr.shape)
```

```python
# reshaping to row vector and column vector (row and column vectors must be 2D vectors, with 1 row in row vector and 1 column in column vector)
row_vector = np.arange(10).reshape(1, -1) # 1 row in row vector
column_vector = np.arange(10).reshape(-1, 1) # 1 column in column vector 
print("Row Vector:")
print(row_vector)
print("Column Vector:")
print(column_vector)
```

```python
# np.newaxis - turn 1D into a row/column vector (same as above, just another method)
arr1 = np.arange(5)
print(arr1.shape)
row_vector = arr1[np.newaxis, :]  # Add new axis (1) for row vector
column_vector = arr1[:, np.newaxis]  # Add new axis (1) for column vector
print("Row Vector:")
print(row_vector)
print(row_vector.shape)
print("Column Vector:")
print(column_vector)
print(column_vector.shape)
```

```python
# flatten() vs ravel()
arr = np.array([[1, 2], [3, 4]])
print("Original Array:")
print(arr)
arr1 = arr.flatten()
print("Flattened Array (flatten()):")
print(arr1) # returns a flattened deep copy of the array, so changes made in the flattened array do NOT affect the original array
arr1[0] = 10
print("Modified Flattened Array (flatten()):")
print(arr1)
print("No Change in Original Array After Flatten Modification:")
print(arr)
arr2 = arr.ravel()
print("Flattened Array (ravel()):")
print(arr2) # returns a flattened shallow copy of the array, so changes made in the ravelled array affect the original array
arr2[0] = 20
print("Modified Ravelled Array (ravel()):")
print(arr2)
print("Changed Original Array After Ravel Modification:")
print(arr)
```

```python
# np.vstack() and np.hstack() (dimesions of both the arrays have to be compatible)
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
np.vstack((a, b))
np.hstack((a, b))
print(np.vstack((a, b))) #vertical stack (no. of columns of both arrays have to be same for vertical stacking)
print(np.hstack((a, b))) #horizontal stack (no. of rows of both arrays have to be same for horizontal stacking)
```

```python
# np. concatenate() with axis
# axis=0 -> vstack, axis=1 -> hstack
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(np.concatenate((a, b), axis=0)) # vertical concatenation 
print(np.concatenate((a, b), axis=1)) # horizontal concatenation
```

```python
# np.where(condition, true, false) 
arr = np.array([[1, 2], [3, 4]])
result = np.where(arr > 2, arr, 0) # condition, if true = original value, if false = 0
print(result)
```
