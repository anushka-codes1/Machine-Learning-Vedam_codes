## Question 1. The Strict Username Validator
Write a function that takes a username string as input and validates it. To be valid, the username must be at
least 5 characters long after removing any leading/trailing spaces, and it must contain only letters and
numbers (no special characters like @, _, or spaces inside). If it is valid, return the clean, strictly lowercase
version of the username. If it is invalid, return the string "Invalid".

```python
def validate_username(username):
    username = username.strip()
    if len(username)<5:
        return "Invalid"
    if not username.isalnum():
        return "Invalid"
    return username.lower()

print(validate_username(" Anushka123 "))
print(validate_username(" Python_Class "))
```

## Question 2. The Spy Code Reverser
You intercept a secret code where words are separated by hyphens (-). To decode it, you must reverse the
order of the words, convert every letter to UPPERCASE, and replace the hyphens with spaces.

```python
def reverse_code(secret_code):
    words = secret_code.split("-")   # Split the string into a list using '-' as the separator.
    words = words[::-1] # Reverse the list
    return " ".join(words).upper()  # Join the reversed list to a string with spaces and convert to uppercase.

secret_code = "charlie-bravo-alpha"
print(reverse_code(secret_code))
```

## Question 3. The Messy Database Cleaner
You have received a raw string of names from a database exported as a semicolon-separated list. The data is
messy, full of random spaces, and has inconsistent casing. Write a function to clean this string so that every
name is properly Title Cased, extra spaces are removed, and the names are joined into a single string
separated by a comma and a space.

```python
def clean_names(raw_names):
    names=raw_names.split(";") # Split the string into a list using ';' as the separator
    cleaned=[] # Create an empty list to store cleaned names

    for name in names:
        cleaned.append(name.strip().title()) # Remove extra spaces and convert the name to Title Case (first letter of each word gets capitalised)

    return ", ".join(cleaned) # Join all cleaned names list into a string using a comma and a space

raw_names = " john doe ; jAnE sMiTh; ALAN tUrInG "
print(clean_names(raw_names))
```

## Question 4. The Multiple Space Eliminator
A user has accidentally typed a sentence with multiple random spaces between words, and the casing is entirely
lowercase. Write code that removes all the extra spaces so there is exactly one space between each word, and
capitalizes only the very first letter of the entire sentence.

```python
def clean_sentence(text):
    words = text.split()
    sentence = " ".join(words)
    return sentence.capitalize()

text = "   python    is   a    very      powerful   language   "
print(clean_sentence(text))
```

## Question 5. The Palindrome Sentence Checker
Determine if a given sentence is a palindrome. To do this correctly using only string methods, remove all spaces
and non-alphanumeric characters, convert everything to lowercase, and check whether it reads the same
forwards and backwards.

```python
def is_palindrome(sentence):
    cleaned=""

    for ch in sentence.strip().lower():
        if ch.isalnum():
            cleaned += ch
    return cleaned == cleaned[::-1]

print(is_palindrome("Was it a car or a cat I saw?"))
print(is_palindrome("Python is fun"))
```

## Question 6. The Hashtag Extractor
Given a string representing a social media post, extract all the hashtags. The extracted hashtags must be
completely lowercase, and the # symbol should be removed. Store them in a list. Ignore any words that are not
hashtags.

```python
def extract_hashtags(post):
    words=post.split() # Coverting the post string to a list named words
    hashtags = [] # An empty list to store the hashtags

    for word in words:
        if word.startswith("#"):
            hashtags.append(word[1:].lower()) # Remove '#' and convert the hashtag to lowercase
    return hashtags

post = "Loving the SUNSET! #Nature #Sunset_Vibes #beautiful"

print(extract_hashtags(post))
```

## Question 7. The File Extension Finder
You are given a single comma-separated string containing various filenames. Extract just the file extensions
(without the dot), ensure there are no duplicates, and return them as a list of strings strictly in UPPERCASE.

```python
def file_extensions(filenames):
    files = filenames.split(",") # Converting the string filenames into a list called files, separating the commas
    extensions=[]

    for file in files:
        file = file.strip() # Removing the extra spaces from the files list
        ext = file.split(".")[-1].upper() # Each word (file) in the files list is converted into another list called ext, separating by the "."
        # So, file = "report.pdf" becomes ['report', 'pdf'] and file.split(".")[-1] gives us the last element: "pdf"

        if ext not in extensions:
            extensions.append(ext)
    return extensions

filenames = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"

print(file_extensions(filenames))
```

## Question 8. The Domain Name Slicer
Given an email address string, extract just the company/domain name — the part immediately after the @
symbol and before the dot suffix like .com or .edu. Convert the extracted domain to Title Case. If the string
does not contain an @, return "Invalid".

```python
def domain_name(email):
    email = email.strip()
    if "@" not in email:
        return "Invalid"
    
    domain= email.split("@")[1] # in email "developer_99@microsoft.com" when we split the email, it becomes ["developer_99" , "microsoft.com"], so [1] gives us the word at index [1] in the list, i.e. "microsoft.com"
    company=domain.split(".")[0]  # in domain "microsoft.com" when we split the domain, it becomes ["microsoft" , "com"], so [0] gives us the word at index [0] in the list, i.e. "microsoft"

    return company.title()

print(domain_name("developer_99@microsoft.com"))
print(domain_name("admin_at_website.org"))
```

## Question 9. The Anagram Detective
Write a function to check if two given strings are anagrams of each other — they contain the exact same letters
in the exact same frequencies, just rearranged. Ignore any spaces and treat uppercase and lowercase letters as
the same.

```python
def is_anagram(str1,str2):
    str1 = str1.replace(" ","").lower()
    str2 = str2.replace(" ","").lower() # makes "Dirty room" - "Dirtyroom"
    return sorted(str1) == sorted(str2)

print(is_anagram("Dormitory", "Dirty room"))
```

## Question 10. The Messy Phone Number Formatter
You are building a CRM system and user-submitted phone numbers are a mess — they contain spaces,
parentheses, periods, and hyphens. Write a function that takes a raw phone number string and cleans it by
removing all non-digit characters. Once cleaned, validate the number:
• If the cleaned string contains exactly 10 digits, format and return it as "(XXX) XXX-XXXX".
• If it contains anything other than 10 digits, return "Invalid Number".

```python
def format_phone(raw_num):
    digits = ""
    for ch in raw_num.strip():
        if ch.isdigit():
            digits += ch
    
    if len(digits)!=10:
        return "Invalid Number"
    
    return "(" + digits[:3] + ") " + digits[3:6] + "-" + digits[6:]

print(format_phone(" +1 (555) 123.4567 "))
print(format_phone("555-987-6543"))
```

## Question 11. The Product SKU Inspector
An inventory system requires product codes (SKUs) to strictly match this format after being cleaned up: exactly
3 uppercase letters, a single hyphen, and exactly 4 digits (e.g. "ABC-1234").
Write a function that receives a messy SKU string. First clean it by stripping leading/trailing spaces and
converting it to uppercase. Then validate it using string methods. Return "Valid: [Cleaned_SKU]" if it
passes, or "Invalid SKU" if it fails any formatting rule.

```python
def validate_sku(sku):
    sku = sku.strip().upper()

    parts = sku.split("-") # In xYz-9876, parts = ["xYz" , "9876"]
    if len(parts)!=2:
        return "Invalid SKU"
    
    letters = parts[0] # "xYz"
    digits = parts[1] # 9876

    if len(letters)!=3:
        return "Invalid SKU"
    if len(digits)!=4:
        "Invalid SKU"
    if not letters.isalpha(): # Checking if xYz contains any other character apart from alphabets
        return "Invalid SKU"
    if not digits.isdigit(): # Checking if 9876 contains any other character apart from digits
        return "Invalid SKU" 
    return "Valid: " + sku

print(validate_sku(" xYz-9876 "))
print(validate_sku("AB-12345"))
print(validate_sku("A@C-1234"))
```
