```python
import pandas as pd
import numpy as np

t = pd.read_csv("Titanic-Dataset.csv")
tip = pd.read_csv("tips.csv")
```

```python
# Which passenger paid the highest fare in each class, and what was their name? (tips-Dataset: t)
tip.groupby("day") # divides each day as a separate dataframe
```

```python
# How many bills on each day
tip.groupby("day").size()
```

```python
# Average bill on each day
tip.groupby("day")['tip'].mean().round(2)
```

```python
# Make a tip_pct column, then average it per day
tip["tip_pct"] = (tip["tip"]/tip["total_bill"])*100
print(tip.groupby("day")['tip_pct'].mean().round(2))
```

```python
# Bigger table, bigger bill. What happens to the tip percentage as the party grows - up, down or flat?
# Compare average tip percentage by party size
size_tip_pct = tip.groupby("size")["tip_pct"].mean().round(2)
print(size_tip_pct) # As the size increases, tip_pct decreases
```

```python
# Average bill for each party size
tip.groupby("size")["total_bill"].mean().round(2)
```

```python
# Titanic Dataset (t): The company report calls first class "somewhat safer". Find the survival rate of each class.
survival_rate = t.groupby("Pclass")["Survived"].mean().round(2) * 100
print(survival_rate)
```

```python
# Average fare paid in each class
fare_paid = t.groupby("Pclass")["Fare"].mean().round(2)
print(fare_paid)
```

```python
# Passengers per boarding port and average age of each group.
print(t.groupby("Embarked").size())
print(t.groupby("Embarked")["Age"].mean().round(2))
```
