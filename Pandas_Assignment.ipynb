## Q8

**In `describe()`, look at the age column. Compare its mean with its 50% value. Which is larger, and what does that suggest about the passengers' ages? ⭐**


```python
# ---- Q8 : your solution ----

mean_age = t["Age"].mean()       # Calculate average age
median_age = t["Age"].median()   # Calculate middle/50% value

print("Mean:", mean_age)
print("50% value:", median_age)
```

## Q9

**(Challenge) The age column has missing values. Using only `shape` and `info()`, work out what percentage of age values are missing. Write the number in a comment.**


```python
# ---- Q9 : your solution ----

total_rows = t.shape[0]       # Total number of rows = 891
age_values = 714              # From t.info(), age has 714 non-null values

missing_age = total_rows - age_values

percentage_missing = (missing_age / total_rows) * 100

print(percentage_missing)

# About 19.87% of the age values are missing.
```
