```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Titanic-Dataset.csv')
mpg = pd.read_csv('mpg.csv')
```

```python
# a plain histogram of Fare, with a size and a title -> histplot
plt.figure(figsize=(7,4)) # figsize is the length and width of the plot generated
sns.histplot(data=df, x='Fare')
plt.title("Fare")
plt.grid()
plt.show()
```

## Histplot -> how spread out is each category?

```python
# Distribution of Fare
plt.figure(figsize=(5,2))
sns.histplot(data=df, x='Fare', bins=25, kde=True, hue='Sex') # bins are the boxes in the grid
plt.title("Fare Distribution")
plt.grid()
plt.show()
```

```python
# Distribution of Age
plt.figure(figsize=(5,2))
sns.histplot(data=df, x='Age', bins=25) # bins are the boxes in the grid
plt.title("Age Distribution")
plt.grid()
plt.show()
```

## Countplot -> How many in each category?

```python
# how many survived and how many died?
plt.figure(figsize=(5,2))
sns.countplot(data=df, x='Survived') # bins are the boxes in the grid
plt.title("Survived")
plt.show()
```

```python
# Plot the distribution of mpg. Is it skewed? If this column had blanks, would you fill with the mean or the median?
plt.figure(figsize=(6,3))
sns.histplot(data=mpg,x='mpg',bins=25)
plt.title("mpg Distribution")
plt.grid()
plt.show()
# The mpg distribution is right-skewed (positively skewed) because most values are concentrated around 15–30 mpg, 
# with a longer tail extending toward higher values (35–47 mpg).
# So, if this column had missing values, I would fill them with the median, not the mean.
# Why?
# The mean is sensitive to extreme values/outliers, such as the relatively high mpg values in the right tail. 
# The median is more robust and better represents the central tendency of a skewed distribution.
```

```python
# How many cars come from each origin? What would that imbalance do to a model trained on this data?
plt.figure(figsize=(6,3))
sns.countplot(data=mpg,x='origin')
plt.title("Cars Count")
plt.show()
# The data is imbalanced.
# If we train a machine-learning model on this data, the model may become biased toward the majority class (USA).
# For example, if we're predicting a car's origin, the model sees:
# Many examples of USA cars → learns USA patterns very well.
# Far fewer examples of Japan/Europe cars → may struggle to recognize them.
```

## Boxplot -> Does a number differ across groups?

```python
plt.figure(figsize=(6,3))
sns.boxplot(data=df,x='Fare')
plt.title("Fare")
plt.show()

# The box is the middle half of the data. The line inside the blue box is the median. The circles beyond the whiskers ( ----| this structure is the whisker) are the outliers.
        #         Outlier
        #            ●
        #            |
        #            |
        # -──────────┤  ← Upper whisker (Q3: 75th percentile)
        # │          │
        # │    BOX   │
        # │     ─────│  ← Median (Q2: 50th percentile)
        # │          │
        # ───────────┤  ← Lower whisker (Q1: 25th percentile)
        #            |
# Q3 − Q1 = IQR (Inter Quartile Range)
```

```python
# Boxplot shows us what impact 'fare' has on 'survived' (higher fare, higher survival rate)
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x='Survived',y='Fare')
plt.title('Survived or not based on Fare')
plt.show()
```

```python
# Boxplot shows us what impact 'age' has on 'survived' (the medians are almost the same so 'age' doesn't have much impact on 'survived or not')
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x='Survived',y='Age')
plt.title('Survived or not based on Age')
plt.show()
```

## Scatterplot -> Do two numbers move together?

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=df,x='Age',y='Fare')
plt.title('Age vs Fare')
plt.show()

# A shapeless cloud, as the correlation between age and fare is just 0.096
```

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=df,x='Age',y='Fare',hue='Sex') # Male and Female will show different colours due to 'hue' parameter
plt.title('Age vs Fare')
plt.show()
```

```python
# Boxplot mpg split by origin, then split by cyliders instead. Which of the two separates the groups more clearly?
plt.figure(figsize=(6,3))
sns.boxplot(data=mpg,x='origin')
plt.title("Origin")
plt.show()
```

```python
plt.figure(figsize=(6,3))
sns.boxplot(data=mpg,x='cylinders')
plt.title("Cylinders")
plt.show()
# Cylinders separates the groups more clearly.
# 1. The first boxplot shows that cars from different origins have different MPG distributions, but there is still considerable overlap between the groups.
# For example:
# USA cars tend to have lower MPG
# Japan and Europe tend to have higher MPG
# But their MPG ranges overlap quite a lot.

# So origin gives some separation, but not perfect separation.

# 2. The second plot shows a much clearer relationship:
# 3–4 cylinders → generally higher MPG
# 6 cylinders → medium MPG
# 8 cylinders → generally lower MPG

# This makes sense because more cylinders generally mean a larger/more powerful engine, which tends to consume more fuel.

# Key takeaway

# Cylinders is a stronger feature for separating cars based on MPG (Miles Per Gallon) than origin.
```
