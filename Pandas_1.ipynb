```python
import pandas as pd
import numpy as np
```

```python
restraunt = {
    'Table_no' : [12, 13, 23, 99],
    'Bill' : [1000, 2100, 2900, 510],
    'Waiter' : ['Navneet', 'Aayush','Manya','Aaryaa']
}

df = pd.DataFrame(restraunt)
print(df)
print(df['Bill']) # This prints the 'Bill' column from the DataFrame
print(type(df['Bill'])) # type is <class 'pandas.Series'>
```

```python
total_bill = [1,2,345,987.0009,87,9876]
print(pd.Series(total_bill)) # pd.Series is used for creating a one-dimensional array-like object from the total_bill list
```

```python
food_count = [45,30,15,60]
food_series= pd.Series(food_count, index = ["Pizza" , "Burger" , "Pasta" , "Fries"])
print(food_series)
print( food_series['Pizza'] + food_series["Burger"] + food_series['Pasta'] + food_series["Fries"] )
```

```python
shop =  {
    "Brand" : ["Puma" , "Adidas" , "Nike" , "Reebook"],
    "Price" : [1000, 5000, 3500 , 1200],
    "In_stock" : [True , False, False, True]
}

print(pd.DataFrame(shop))
```
